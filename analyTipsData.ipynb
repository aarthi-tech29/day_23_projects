### Project 2: Analyzing Tips Data 
#### Objective: Explore the relationships between tip amounts, total bill, and dining details.

##### Step 1. Load the Dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
```

##### Step 2. Relational Plot

```python
sns.relplot(
    x='total_bill',
    y='tip',
    hue='day',
    style='time',
    data=tips
)
plt.title("Total Bill vs Tip by Day and Time")
plt.show()

# Shows how tip increases as total bill increases.
# Color (hue) shows which day customers tipped more.
# Style (time) shows differences between Lunch and Dinner.
# Demonstrates the power of relplot to combine multiple variables in one chart.
```

##### Step 3. Categorical Analysis

```python
# Tip Distribution by Day
sns.violinplot(x='day', y='tip', data=tips)
plt.title("Tip Distribution Across Days")
plt.show()

# Violin plot shows full distribution, not just the mean.
# Helps compare how tipping behavior changes by weekday.
# can see which day has higher or more spread-out tips.
```

##### Step 4. Heatmap for Correlation

```python
# Compute and Visualize the Correlation Matrix
corr = tips.corr(numeric_only=True)

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap of Numerical Features")
plt.show()

# Shows numerical relationships clearly.
# For example, total_bill and tip will have a strong positive correlation.
# Helps identify which variables influence each other.

# Total bill and tip (0.68) have the strongest positive relationship.
# → Higher bills generally lead to higher tips.

# Total bill and size (0.60) have a moderately strong relationship.
# → Bigger groups usually spend more.

# Tip and size (0.49) have the weakest relationship.
# → Tips do not depend heavily on the number of people.

# There are all 1.0 because every variable is perfectly correlated with itself.That’s why they are light colored.
```
