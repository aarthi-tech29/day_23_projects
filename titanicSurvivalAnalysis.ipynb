### Project 1: Titanic Survival Analysis 
#### Objective: Use visualization to explore the factors influencing passenger survival on the Titanic.

##### Step 1. Load the Dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset('titanic')
```

##### Step 2. Univariate Analysis

```python
# Distribution of Age
sns.histplot(data=titanic, x='age')
plt.title("Distribution of Passenger Age")
plt.show()

# sns.histplot() to visualize how passenger ages are distributed
# Shows common age ranges
# Helps identify skewness or missing values
```

##### Step 3. Bivariate & Categorical Analysis

```python
# Survival by Class
sns.barplot(x='class', y='survived', data=titanic)
plt.title("Survival Rate by Passenger Class")
plt.show()

# First-class passengers had a much higher survival rate
# Third-class passengers had the lowest
```

```python
# Survival by Gender
sns.countplot(x='sex', hue='survived', data=titanic)
plt.title("Survival Count by Gender")
plt.show()

# Shows number of survivors vs non-survivors
# Confirms the “women and children first” rule
# Women had a significantly higher survival rate
```

##### Step 4. Multivariate Analysis

```python
# Fare, Age, and Survival
sns.scatterplot(x='age', y='fare', hue='survived', data=titanic)
plt.title("Age vs Fare vs Survival")
plt.show()

# Helps see whether higher fare (wealthier passengers) had a better chance of survival
# Allows spotting age patterns among survivors
# Shows how multiple factors interact
```
