### Project 2: Student Test Score Analysis 
#### Objective: Explore a student test score dataset to find relationships between scores and demographic factors. 
#### Technique: Use boxplot to compare scores by gender, violinplot to see score distributions for different study groups, and scatterplot to visualize the relationship between scores on two different subjects.

##### Step 1: Load the dataset

```python
import pandas as pd


df = pd.read_csv("student_scores.csv")

# To clean the dataset, especially the wrong value "ninety" in the row 3 in science_score
df["science_score"] = pd.to_numeric(df["science_score"], errors="coerce")
df["science_score"].fillna(df["science_score"].mean())

# Convert everything to numeric
# Replace NaN with the mean value

df.head()
```

##### Step 2. Boxplot — Compare Scores by Gender

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x="gender", y="math_score", data=df)
plt.title("Math Score Distribution by Gender")
plt.show()

# A boxplot helps you compare the score distribution between genders.
# Median (middle line) → typical score
# Box → middle 50% of scores
# Whiskers → overall range
# Dots → outliers
```

##### Step 3. Violinplot — Score Distribution by Study Group

```python
sns.violinplot(x="study_group", y="reading_score", data=df)
plt.title("Reading Score Distribution by Study Group")
plt.show()

# # A violinplot shows the full score distribution (shape + density).
# # Where most students cluster
# # Whether the group has high variance
# # If there are multiple peaks (bimodal distribution)
# # One bulge = unimodal
# Two bulges/humps = bimodal
# Many humps = multimodal
# Which study group performs consistently?
# Which group has wide variation?
```

##### Step 4. Scatterplot — Relationship Between Two Scores

```python
sns.scatterplot(x="math_score", y="science_score", data=df)
plt.title("Math vs Science Score Relationship")
plt.show()

# A scatterplot shows how one subject score relates to another.
# Positive relationship: high math ⇒ high science
# Negative relationship: high math ⇒ low science
# No relationship: random dots
# Do students strong in math also score well in science?
```
