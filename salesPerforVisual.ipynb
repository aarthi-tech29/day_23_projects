### Project 1: Sales Performance Visualization 
#### Objective: Use a sales dataset to visualize sales performance. 
#### Technique: Plot a lineplot to show sales trend over time, a barplot to compare sales across different regions, and a heatmap of a pivot table to show sales by month and product category. 

##### Step 1: Load the dataset

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sales_df = pd.DataFrame({
    'date': ['2023-01', '2023-02', '2023-03'],
    'region': ['North', 'South', 'East'],
    'sales': [12000, 15000, 13000],
    'month': ['Jan', 'Feb', 'Mar'],
    'category': ['Electronics', 'Clothing', 'Grocery']
})

print(sales_df)
```

##### Step 2. Line Plot – Sales Trend Over Time

```python

sns.lineplot(x='date', y='sales', data=sales_df)
plt.title("Sales Trend Over Time")
plt.show()

# Use a lineplot to show how sales change over time.
# Helps identify increasing or decreasing trends.
# Shows seasonal patterns (e.g., higher sales in December).
# Useful for forecasting and decision-making.
```

##### Step 3. Bar Plot – Sales by Region

```python
sns.barplot(x='region', y='sales', data=sales_df)
plt.title("Sales Performance by Region")
plt.show()

# Use a barplot to compare total sales across different regions.
# Shows which region is performing the best or worst.
# Highlights regional variation in customer demand.
# Helps in region-specific strategy planning.
```

##### Step 4. Heatmap – Sales by Month and Product Category

```python
# create a pivot table
pivot = sales_df.pivot_table(
    values='sales',
    index='month',
    columns='category',
    aggfunc='sum'
)

# visualize using a heatmap
sns.heatmap(pivot, annot=True)
plt.title("Monthly Sales by Product Category")
plt.show()

# Shows which months have high or low sales.
# Compares performance across product categories.
# Helps identify best-selling categories and seasonal demand patterns.

# Best performing month-category combination
# Clothing in February (15000) → highest sales

# Lowest performing category-month combination
# Electronics in January (12000)

# Grocery in March performed moderately (13000)
# The color gradient helps see the trend:

# Lighter → better sales
# Darker → lower sales

# Lineplot → overall sales trend
# Barplot → region-wise comparison
# Heatmap → month-category relationship
```
